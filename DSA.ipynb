Types of Time Complexity (Brief Summary)

Time Complexity	Name	Explanation	Example

O(1)	Constant	Execution time remains the same regardless of input size.	Accessing an array element by index.

O(log n)	Logarithmic	Input size is reduced by half (or a fixed factor) at each step.	Binary Search.

O(n)	Linear	Execution time increases proportionally with input size.	Traversing an array using a loop.

O(n log n)	Linearithmic	Combines linear and logarithmic operations.	Merge Sort, Heap Sort.

O(n²)	Quadratic	Usually involves two nested loops.	Bubble Sort, Selection Sort.

O(n³)	Cubic	Usually involves three nested loops.	Basic matrix multiplication.

O(2ⁿ)	Exponential	Work doubles with every additional input element.	Naive recursive Fibonacci.

O(n!)	Factorial	Generates all possible arrangements or permutations of input elements.	Brute-force Traveling Salesman Problem.
